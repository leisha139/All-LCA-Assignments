```python
import pandas as pd

# Step 1: Create sample student data
data = {
    "Name": ["Rahul", "Priya", "Amit", "Rahul", "Neha"],
    "Marks": [85, None, 72, 85, 90],
    "City": ["Pune", "Mumbai", "Pune", "Pune", None]
}

df = pd.DataFrame(data)

# Step 2: Display the original data
print("Original Data:")
print(df)

# Step 3: Explore the dataset
print("\nFirst five rows:")
print(df.head())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
print(df.describe())

# Step 4: Check for missing values
print("\nMissing values:")
print(df.isnull().sum())

# Step 5: Fill missing values
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["City"] = df["City"].fillna(df["City"].mode()[0])

# Step 6: Remove duplicate records
df = df.drop_duplicates()

# Step 7: Remove an irrelevant column (if present)
# df = df.drop(columns=["ColumnName"])

# Step 8: Select specific columns
print("\nSelected columns:")
print(df.loc[:, ["Name", "Marks"]])

# Step 9: Filter students scoring 80 or more
print("\nStudents scoring 80 or more:")
print(df.loc[df["Marks"] >= 80])

# Step 10: Display the cleaned data
print("\nCleaned Data:")
print(df)

# Step 11: Save the cleaned data
df.to_csv("cleaned_students.csv", index=False)
```
