```python
import re

def check_allowed_chars(input_string):

    pattern = r'^[a-zA-Z0-9]+$'
    
    if re.match(pattern, input_string):
        return True
    else:
        return False


test_str_1 = "Python2026"
test_str_2 = "Hello, World!"

print(f"'{test_str_1}' is valid: {check_allowed_chars(test_str_1)}")
print(f"'{test_str_2}' is valid: {check_allowed_chars(test_str_2)}")
```
