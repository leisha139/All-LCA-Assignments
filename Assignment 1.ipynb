```python
student = {"name": "Alice", "age": 15, "grade": "10th"}
print("Initial Dictionary:", student)


student["subject"] = "Computer Science"
print("After Adding/Appending:", student)

removed_value = student.pop("age")
print(f"After Removing 'age' (Removed: {removed_value}):", student)

my_tuple = (10, 20, 30)
print("\nInitial Tuple:", my_tuple)

temp_list = list(my_tuple)

temp_list.append(40)

temp_list.remove(10)

my_tuple = tuple(temp_list)
print("Modified Tuple (via list conversion):", my_tuple)
```
